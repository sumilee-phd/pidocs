## 2.2 반복문

[본문 보기](https://pidocs.aifusionlab.org/ai-python/ch02/2-2/)

```python
readings = [-63, -53, -54, -52, -55]   # 신호 세기(dBm)
total = 0
for value in readings:
    total = total + value
print("합계", total)
print("평균", total / len(readings))
```

```python
for i in range(1, 4):
    print(i, "회차 측정")
```

```python
count = 0
total = 0
for value in readings:
    if value < -60:          # 약한 신호는 집계에서 제외
        print("제외", value)
        continue
    total += value
    count += 1
print("평균", total / count, "(", count, "건 )")
```

```python
target = -54
for i in range(len(readings)):
    if readings[i] == target:
        print(i + 1, "번째 측정에서 발견")
        break
```

```python
count = 0
while count < 3:
    print("측정", count + 1)
    count += 1
```

## 2.3 함수

[본문 보기](https://pidocs.aifusionlab.org/ai-python/ch02/2-3/)

```python
def judge_light(bri):
    """조도값(0~100)을 밝기 단계 문자열로 반환한다."""
    if bri < 30:
        return "어두움"
    elif bri < 70:
        return "보통"
    else:
        return "밝음"

result = judge_light(63)
print(result)
print(judge_light(12), judge_light(88))
```

```python
def judge_light2(bri, dark=30, bright=70):
    if bri < dark:
        return "어두움"
    elif bri < bright:
        return "보통"
    return "밝음"

print(judge_light2(63))
print(judge_light2(63, dark=70, bright=90))
```

## 2.4 실습: 조도값 판정 함수

[본문 보기](https://pidocs.aifusionlab.org/ai-python/ch02/2-4/)

```python
def judge_light(bri):
    """조도값을 판정한다. 0~100 범위 밖이면 "측정 오류"를 반환한다."""
    if bri < 0 or bri > 100:
        return "측정 오류"
    if bri < 30:
        return "어두움"
    if bri < 70:
        return "보통"
    return "밝음"

for v in [-1, 0, 29, 30, 69, 70, 100, 101]:
    print(v, judge_light(v))
```

```python
assert judge_light(29) == "어두움"
assert judge_light(30) == "보통"
assert judge_light(70) == "밝음"
assert judge_light(101) == "측정 오류"
print("경계값 시험 통과")
```

```python
day_log = [12, 25, 41, 58, 66, 73, 88, 95, 71, 47, 22, 9]
dark = normal = bright = error = 0
for v in day_log:
    r = judge_light(v)
    if r == "어두움":
        dark += 1
    elif r == "보통":
        normal += 1
    elif r == "밝음":
        bright += 1
    else:
        error += 1
print("어두움", dark, "보통", normal, "밝음", bright, "오류", error)
```
