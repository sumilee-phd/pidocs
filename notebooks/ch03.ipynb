```python
def judge_light(bri):
    if bri < 0 or bri > 100:
        return "측정 오류"
    if bri < 30:
        return "어두움"
    if bri < 70:
        return "보통"
    return "밝음"

day_log = [12, 25, 41, 58, 66, 73, 88, 95, 71, 47, 22, 9]
counts = {}
for v in day_log:
    r = judge_light(v)
    counts[r] = counts.get(r, 0) + 1
print(counts)
```

```python
records = [
    {"node": "S02", "zone": "D1", "rssi": -63},
    {"node": "S03", "zone": "D3", "rssi": -65},
    {"node": "N6", "zone": "D4", "rssi": -51},
    {"node": "S01", "zone": "D2", "rssi": -48},
]
for r in records:
    print(r["zone"], r["node"], r["rssi"])
strongest = max(records, key=lambda r: r["rssi"])
print("가장 강한 신호:", strongest)
```

## 3.2 파일 입출력과 CSV

[본문 보기](https://pidocs.aifusionlab.org/ai-python/ch03/3-2/)

```python
import csv

rows = [["zone", "node", "rssi"], ["D1", "S02", -63], ["D3", "S03", -65], ["D4", "N6", -51]]
with open("sample.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerows(rows)

with open("sample.csv", encoding="utf-8") as f:
    print(f.read())
```

```python
import csv

zone_sum = {}
zone_cnt = {}
with open("class_rssi_20261104.csv", encoding="utf-8") as f:
    for row in csv.DictReader(f):
        z = row["zone"]
        zone_sum[z] = zone_sum.get(z, 0) + int(row["rssi"])
        zone_cnt[z] = zone_cnt.get(z, 0) + 1

print("지점  건수  평균(dBm)")
for z in sorted(zone_cnt):
    print(z, f"{zone_cnt[z]:5d}", f"{zone_sum[z] / zone_cnt[z]:9.1f}")
```

## 3.3 실습: 서울 날씨 자료 첫 분석

[본문 보기](https://pidocs.aifusionlab.org/ai-python/ch03/3-3/)

```python
import csv

stats = {}
with open("seoul_weather_2021_2025.csv", encoding="utf-8") as f:
    for row in csv.DictReader(f):
        year = row["date"][:4]
        s = stats.setdefault(year, {"n": 0, "temp": 0.0, "rain_days": 0, "precip": 0.0})
        p = float(row["precip_mm"])
        s["n"] += 1
        s["temp"] += float(row["temp_mean"])
        s["precip"] += p
        if p >= 1.0:
            s["rain_days"] += 1

print("연도  일수  평균기온  강수일수  연강수량")
for year, s in stats.items():
    print(year, f"{s['n']:4d}", f"{s['temp'] / s['n']:8.1f}", f"{s['rain_days']:8d}", f"{s['precip']:9.1f}")
```
